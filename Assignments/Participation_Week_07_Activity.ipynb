```python
df = pd.read_csv('/content/drive/MyDrive/MSSP607/Modules/Week_7 folder/InternetSales.csv', encoding='latin-1')
```

```python
df.head()
```

```python
# Extract Year
df['OrderDate'] = pd.to_datetime(df['OrderDate'])
df['OrderYear'] = df['OrderDate'].dt.year

# Group by year and sum sales
yearly_sales = df.groupby('OrderYear')['SalesAmount'].sum().reset_index()

# Create the bar chart
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.bar(yearly_sales['OrderYear'], yearly_sales['SalesAmount'])
plt.xlabel('Year')
plt.ylabel('Total Sales Amount')
plt.title('Yearly Sales')
plt.xticks(yearly_sales['OrderYear'])
plt.grid(axis='y')

# Save the figure before showing it
plt.savefig('yearly_sales_chart.pdf', bbox_inches='tight')

plt.show()
```
