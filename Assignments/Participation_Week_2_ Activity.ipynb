```python
letters = ['a','b','c','d']
letters[int(int('3'*2)//11)]
print(letters[int(int('3'*2)//11)])
print("Explanation: '3'*2 = '33' -> int('33') = 33 -> 33//11 = 3 -> letters[3] = 'd'")
```

```python
letters[-1]
print(letters[-1])
print("Explanation: Index -1 means the last element, so it returns 'd'")
```

```python
letters[:2]
print(letters[:2])
print("Explanation: [:2] slices from start up to index 2 (exclusive), so it returns ['a', 'b']")
```

```python
def list_to_string(spam):
    if not spam:
        return ''
    elif len(spam) == 1:
        return spam[0]
    elif len(spam) == 2:
        return f"{spam[0]} and {spam[1]}"
    else:
        return ', '.join(spam[:-1]) + f", and {spam[-1]}"
spam = ['apples', 'bananas', 'tofu', 'cats']
result = list_to_string(spam)
print(result)
```
