```python
import pandas as pd
import os
```

```python
Apple = pd.read_csv('/content/MSSP607/Data/aapl.csv')
Facebook = pd.read_csv('/content/MSSP607/Data/fb.csv')
Amazon = pd.read_csv('/content/MSSP607/Data/amzn.csv')
Google = pd.read_csv('/content/MSSP607/Data/goog.csv')
Netflix = pd.read_csv('/content/MSSP607/Data/nflx.csv')
```

```python
Apple['ticker'] = 'AAPL'
Facebook['ticker'] = 'FB'
Amazon['ticker'] = 'AMZN'
Google['ticker'] = 'GOOG'
Netflix['ticker'] = 'NFLX'
```

```python
all_stocks = pd.concat([Apple, Facebook, Amazon, Google, Netflix], ignore_index=True)
all_stocks.head()
```

```python
all_stocks.to_csv('/content/MSSP607/Data/faanf.csv', index=False)
```

```python
top7_volume = all_stocks.nlargest(7, 'volume')
top7_volume
```

```python
melted_stocks = all_stocks.melt(
    id_vars=['date', 'ticker'],
    var_name='variable',
    value_name='value'
)

melted_stocks.head()
```
