```python
# Question b
x = "I'm in the GLOBAL scope"

def test_scope():
    y = "I'm in the LOCAL scope"
    print("Inside the function:", y)

test_scope()
print("Outside the function:", x)

print("There is one global scope per program. The number of local scope is not fixed and can be many.")
```

```python
# Question c
def local_scope_demo():
    z = 42
    print("Inside function, z =", z)

local_scope_demo()

try:
    print("Outside function, z =", z)
except NameError as e:
    print("Error:", e)

print("Answer: variables in a local scope are destroyed when the function finishes.")
```

```python
# Question d
def safe_divide(x, y):
    try:
        result = x / y
        return result
    except ZeroDivisionError:
        return "Cannot divide by zero!"

print("10 / 2 =", safe_divide(10, 2))
print("10 / 0 =", safe_divide(10, 0))

print("Answer: Errors can be handled with try and except statements.")
```

```python
# The Collatz Sequence Project
def collatz(number):
    if number % 2 == 0:
        result = number // 2
    else:
        result = 3 * number + 1

    print(result)
    return result

try:
    user_input = int(input("Enter number:\n"))
    while user_input != 1:
        user_input = collatz(user_input)
except ValueError:
    print("You must enter a valid integer.")
```
