```python
# Question b
indonesia_quakes = df[df["parsed_place"].str.contains("Indonesia", case=False, na=False)]
tsunami_percentage = (indonesia_quakes['tsunami'] == 1).mean() * 100
print(f"Percentage of earthquakes in Indonesia with tsunamis: {tsunami_percentage:.2f}%")
```

```python
#Question d
ring_of_fire_locations = [
    'Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica', 'Guatemala', 'Mexico',
    'Japan', 'Philippines', 'Indonesia', 'New Zealand', 'Antarctica', 'Canada',
    'Fiji', 'Alaska', 'Washington', 'California', 'Russia', 'Taiwan', 'Tonga',
    'Kermadec Islands'
]
pattern = '|'.join([re.escape(location) for location in ring_of_fire_locations])

df['ring_of_fire'] = df['parsed_place'].str.contains(pattern, case=False, na=False)

df.loc[df['parsed_place'].str.contains('New Mexico', case=False, na=False), 'ring_of_fire'] = False

print(df[['parsed_place', 'ring_of_fire']].head())
```

```python
# Question e
ring_of_fire_counts = df['ring_of_fire'].value_counts()

print("Number of earthquakes in and outside the Ring of Fire:")
print(ring_of_fire_counts)
```

```python
# Question f
ring_of_fire_tsunamis = df[(df['ring_of_fire'] == True) & (df['tsunami'] == 1)]
tsunami_count_ring_of_fire = len(ring_of_fire_tsunamis)
print(f"Number of tsunamis in the Ring of Fire: {tsunami_count_ring_of_fire}")
```
